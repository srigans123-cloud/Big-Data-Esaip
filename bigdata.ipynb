```python
from decimal import Decimal, getcontext

getcontext().prec = 50

# Decimal units: powers of 1000
DECIMAL = {
    "B": Decimal("1"),
    "KB": Decimal("1000"),
    "MB": Decimal("1000000"),
    "GB": Decimal("1000000000"),
    "TB": Decimal("1000000000000"),
    "PB": Decimal("1000000000000000")
}

# Binary units: powers of 1024
BINARY = {
    "B": Decimal("1"),
    "KiB": Decimal("1024"),
    "MiB": Decimal("1048576"),
    "GiB": Decimal("1073741824"),
    "TiB": Decimal("1099511627776"),
    "PiB": Decimal("1125899906842624")
}

ALL_UNITS = {**DECIMAL, **BINARY}


def convert(value, from_unit, to_unit):
    value = Decimal(str(value))

    if from_unit not in ALL_UNITS:
        raise ValueError("Invalid FROM unit")

    if to_unit not in ALL_UNITS:
        raise ValueError("Invalid TO unit")

    # Convert to bytes first
    bytes_value = value * ALL_UNITS[from_unit]

    # Convert bytes to target unit
    result = bytes_value / ALL_UNITS[to_unit]

    return result


def calculator():
    print("=" * 50)
    print("       BYTE CONVERSION CALCULATOR")
    print("=" * 50)

    print("\nAvailable units:")
    print("B, KB, MB, GB, TB, PB")
    print("KiB, MiB, GiB, TiB, PiB")

    value = input("\nEnter value: ")
    from_unit = input("Convert FROM: ")
    to_unit = input("Convert TO: ")

    try:
        result = convert(value, from_unit, to_unit)

        print("\n" + "-" * 50)
        print(f"{value} {from_unit} = {result} {to_unit}")
        print("-" * 50)

    except Exception as e:
        print("Error:", e)


calculator()
```
